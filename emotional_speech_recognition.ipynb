```python
import numpy as np
import librosa
import soundfile as sf
from pathlib import Path

# Your dataset path as a Path object
pic_path = Path('/content/drive/MyDrive/speech emotion')

# Emotion mapping from RAVDESS filename
emotion_map = {
    '01': 'neutral',
    '02': 'calm',
    '03': 'happy',
    '04': 'sad',
    '05': 'angry',
    '06': 'fearful',
    '07': 'disgust',
    '08': 'surprised'
}

def extract_features(file_path):
    try:
        audio, sample_rate = sf.read(file_path)
        if len(audio.shape) > 1:  # Convert stereo to mono if needed
            audio = np.mean(audio, axis=1)
        mfccs = librosa.feature.mfcc(y=audio.astype(float), sr=sample_rate, n_mfcc=40)
        mfccs_processed = np.mean(mfccs.T, axis=0)
        return mfccs_processed
    except Exception as e:
        print(f"❌ Error encountered while parsing file: {file_path}\n{e}")
        return None

features = []

# Loop through all .wav files recursively using pathlib
for wav_file in pic_path.rglob('*.wav'):
    data = extract_features(wav_file)
    if data is not None:
        emotion_code = wav_file.name.split('-')[2]
        emotion = emotion_map.get(emotion_code)
        if emotion:
            features.append([data, emotion])

print(f"\n✅ Successfully extracted features from {len(features)} files.")
```

```python
import pandas as pd
df = pd.DataFrame(features, columns=['feature', 'emotion'])
df
```

```python
X = np.array(df['feature'].tolist())
y = np.array(df['emotion'].tolist())
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
model = MLPClassifier(hidden_layer_sizes=(256, 128, 64), activation='relu', solver='adam', max_iter=500)
model.fit(X_train, y_train)

```
